```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-3, 3, 400)
rho = np.exp(-z**2)

plt.figure(figsize=(6, 4))
plt.plot(z, rho)
plt.xlabel('z')
plt.ylabel(r'$\rho = e^{-z^2}$')
plt.title(r'$\rho(z) = e^{-z^2}$')
plt.grid(True, alpha=0.3)
plt.axhline(0, color='k', lw=0.5)
plt.axvline(0, color='k', lw=0.5)
plt.tight_layout()
plt.show()
```
